## Initialize an empty phone book dictionary

```python
phone_book = {}
```

## Function to add a contact

```python
def add_contact(name, phone, email):
    phone_book[name] = {'phone':phone, 'email':email}
    print(f"Contact '{name}' added successfully!")
```

```python
add_contact(name='Zahra', phone='0913', email='za@gmail.com')
```

```python
phone_book
```

## Function to search for a contact

```python
phone_book.get('Zahra')
```

```python
def search_contact(name):
    contact = phone_book.get(name)
    if contact:
        print(f"Name: {name}")
        print(f"Phone: {contact['phone']}")
        print(f"Email: {contact['email']}")
    else:
        print(f"Contact '{name}' not found.")
```

```python
search_contact('Zahra')
```

## Function to delete a contact

```python
def delet_contact(name):
    if name in phone_book:
        del phone_book[name]
    else:
        print(f"Contact '{name}' not found.")
```

```python
delet_contact('Zahra')
```

```python
phone_book
```

## Function to display all contacts

```python
def show_all_contacts():
    if phone_book:
        print("All contacts: ")
        for name, info in phone_book.items():
            print(f"Name: {name} - Phone: {info['phone']} - Email: {info['email']}")
    else:
        print("The phone book is empty.")
```

```python
show_all_contacts()
```

## Console menu to interact with the phone book

```python
def main():
    while True:
        print("\nPhone Book")
        print("1. Show All Contacts")
        print("2. Add New Contact")
        print("3. Search for a Contact")
        print("4. Delete a Contact")
        print("5. Exit")

        choice = input("Select an option: ")

        if choice == '1':
            show_all_contacts()
        elif choice == '2':
            name = input("Enter name: ")
            phone = input("Enter phone: ")
            email = input("Enter email: ")
            add_contact(name, phone, email)
        elif choice == '3':
            name = input("Enter name to search: ")
            search_contact(name)
        elif choice == '4':
            name = input("Enter name to delete: ")
            delete_contact(name)
        elif choice == '5':
            print("Exiting the program.")
            break
        else:
            print("Invalid option. Please try again.")
```

## Run the main function to start the console application

```python
if __name__ == "__main__":
    main()
```
