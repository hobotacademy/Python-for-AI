```python
import matplotlib.pyplot as plt
```

```python
plt.plot([1, 2, 3, 4], [1, 4, 9, 16])
plt.title("Line Plot")
plt.xlabel("Number")
plt.ylabel("power-2")
plt.show()
```

```python
fig, ax = plt.subplots()
plt.plot([1, 2, 3, 4], [1, 4, 9, 16])
ax.set_title("Ant")
plt.show()
```

```python
x = [0, 1, 2, 3, 4]
y = [0, 1, 4, 9, 16]
```

```python
plt.plot(x, y, color='blue', linestyle='--', marker='x', label='y=x^2')
plt.xlabel("Number")
plt.ylabel("power-2")
plt.title("Line Plot")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# scatter
import numpy as np
x = np.random.rand(50)
y = np.random.rand(50)
```

```python
sizes = np.random.rand(50) * 100
colors = np.random.rand(50)
plt.scatter(x, y, s=sizes, c=colors, cmap='Blues', alpha=0.5)
plt.colorbar()
plt.show()
```

```python
# bar plot
categories = ['A', 'B', 'C', 'D']
values = [5, 7, 3, 8]
```

```python
plt.bar(categories, values, color='green', alpha=0.7)
plt.show()
```

```python
# histogram
data = np.random.randn(1000)
```

```python
plt.hist(data, bins=100, color='purple', edgecolor='white')
plt.show()
```

```python
sizes = [20, 30, 25, 25]
labels = ['A', 'B', 'C', 'D']
explode = (0, 0.2, 0, 0)
```

```python
plt.pie(sizes, labels=labels, explode=explode, autopct='%1.3f%%')
plt.show()
```
