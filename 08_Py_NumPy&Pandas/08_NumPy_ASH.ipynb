### Normal Distribution
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">

Generate random numbers with a bell-curve (Gaussian) distribution.
</div>

`np.random.randn` (mean = 0, std deviation = 1):

```python
import matplotlib.pyplot as plt

# Uniform Distribution
def plot_uniform(a, b, size=1000):
    data = np.random.uniform(a, b, size)
    plt.hist(data, bins=30, density=True, alpha=0.6, color='blue', edgecolor='black')
    plt.title(f'Uniform Distribution (a={a}, b={b})')
    plt.xlabel('Value')
    plt.ylabel('Density')
    plt.show()

# Normal Distribution
def plot_normal(mu, sigma, size=1000):
    data = np.random.normal(mu, sigma, size)
    plt.hist(data, bins=30, density=True, alpha=0.6, color='green', edgecolor='black')
    plt.title(f'Normal Distribution (mu={mu}, sigma={sigma})')
    plt.xlabel('Value')
    plt.ylabel('Density')
    plt.show()

# Exponential Distribution
def plot_exponential(lam, size=1000):
    data = np.random.exponential(1/lam, size)  # Scale = 1/lambda
    plt.hist(data, bins=30, density=True, alpha=0.6, color='red', edgecolor='black')
    plt.title(f'Exponential Distribution (lambda={lam})')
    plt.xlabel('Value')
    plt.ylabel('Density')
    plt.show()

# Plotting the distributions
```

```python
plot_uniform(0, 10)  # Uniform distribution with range [0, 10]
```

```python
plot_normal(0, 1)    # Normal distribution with mu=0, sigma=1
```

```python
plot_exponential(2)  # Exponential distribution with lambda=2
plot_exponential(100)
```

## 13. Mathematical Functions in NumPy

**Table of Functions**<br><br>
| Function         | Description                  | Example                             |
|------------------|------------------------------|-------------------------------------|
| <code>np.add</code>   | Element-wise addition         | <code>np.add([1, 2], [3, 4]) → [4, 6]</code>    |
| <code>np.subtract</code> | Element-wise subtraction      | <code>np.subtract([5, 7], [2, 3]) → [3, 4]</code> |
| <code>np.multiply</code> | Element-wise multiplication    | <code>np.multiply([2, 3], [4, 5]) → [8, 15]</code> |
| <code>np.divide</code>   | Element-wise division          | <code>np.divide([8, 9], [2, 3]) → [4, 3]</code>   |
| <code>np.power</code>    | Element-wise exponentiation    | <code>np.power([2, 3], [3, 2]) → [8, 9]</code>    |
| <code>np.mod</code>      | Remainder of division         | <code>np.mod([10, 15], [3, 5]) → [1, 0]</code>     |

```python
array1 = np.array([1, 2, 3])
array2 = np.array([4, 5, 6])
result = np.add(array1, array2)
print(result)
```

```python
array1 = np.array([10, 20, 30])
array2 = np.array([5, 10, 15])
result = np.subtract(array1, array2)
print(result)
```

```python
array1 = np.array([2, 3])
array2 = np.array([4, 5])
result = np.multiply(array1, array2)
print(result)
```

```python
array1 = np.array([8, 9])
array2 = np.array([2, 3])
result = np.divide(array1, array2)
print(result)
```

```python
array1 = np.array([2, 3])
array2 = np.array([3, 2])
result = np.power(array1, array2)
print(result)
```

```python
array1 = np.array([10, 15])
array2 = np.array([3, 5])
result = np.mod(array1, array2)
print(result)
```

## 14. Statistical Functions

These functions are useful for data analysis:<br><br>

| Function         | Description              | Example                              |
|------------------|--------------------------|--------------------------------------|
| <code>np.sum</code>    | Sum of all elements       | <code>np.sum([1, 2, 3]) → 6</code>         |
| <code>np.mean</code>   | Mean (average)           | <code>np.mean([1, 2, 3]) → 2</code>        |
| <code>np.median</code> | Median                   | <code>np.median([1, 2, 3, 4]) → 2.5</code> |
| <code>np.std</code>    | Standard deviation       | <code>np.std([1, 2, 3]) → 0.816</code>     |
| <code>np.var</code>    | Variance                 | <code>np.var([1, 2, 3]) → 0.666</code>     |
| <code>np.min</code>    | Minimum value            | <code>np.min([1, 2, 3]) → 1</code>         |
| <code>np.max</code>    | Maximum value            | <code>np.max([1, 2, 3]) → 3</code>         |

```python
array = np.array([1, 2, 3, 4])
result = np.sum(array)
print(result)
```

```python
array = np.array([10, 20, 30])
result = np.mean(array)
print(result)
```

```python
array = np.array([1, 2, 3, 4])
result = np.median(array)
print(result)
```

```python
array = np.array([1, 2, 3])
result = np.std(array)
print(result)
```

```python
array = np.array([1, 2, 3])
result = np.var(array)
print(result)
```

```python
array = np.array([3, 1, 4, 1, 5])
result = np.min(array)
print(result)
```

```python
result = np.max(array)
print(result)
```

## 15. Linear Algebra Operations

For working with matrices and vectors:<br>

| Function           | Description                          | Example                                          |
|--------------------|--------------------------------------|--------------------------------------------------|
| <code>np.dot</code>   | Dot product of vectors or matrices   | <code>np.dot([1, 2], [3, 4]) → 11</code>          |
| <code>np.matmul</code> | Matrix multiplication                | <code>np.matmul([[1, 2]], [[3], [4]]) → [[11]]</code> |
| <code>np.linalg.inv</code> | Matrix inverse                    | <code>np.linalg.inv([[1, 2], [3, 4]])</code>      |
| <code>np.linalg.det</code> | Determinant of a matrix           | <code>np.linalg.det([[1, 2], [3, 4]]) → -2</code>  |
| <code>np.transpose</code> | Transpose of an array            | <code>np.transpose([[1, 2], [3, 4]]) → [[1, 3], [2, 4]]</code> |

```python
vector1 = np.array([1, 2])
vector2 = np.array([3, 4])
result = np.dot(vector1, vector2)
print(result)
```

```python
np.matmul(vector1, vector2)
```

```python
A = np.array([[2, 3], [1, 4]])
```

```python
A_1 = np.linalg.inv(A)
A_1
```

```python
np.linalg.det(A)
```

```python
matrix = np.array([[1, 2], [3, 4]])
transpose = np.transpose(matrix)
print(transpose)
```
