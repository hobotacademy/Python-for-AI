### 2.1. `if`, `else`

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
In Python, conditions are defined using the <code>if</code> keyword. If the condition after <code>if</code> evaluates to <code>True</code>, the code inside the <code>if</code> block is executed.<br><br>

```python
if condition:
    # Code to execute if the condition is true
```</div>


<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
The <code>else</code> statement is used to specify an action to be taken if the <code>if</code> condition is <code>False</code>.<br><br>

```python
if condition:
    # Code to execute if the condition is true
else:
    # Code to execute if the condition is false
```
</div>

````python
age = int(input("Enter your age: "))
````

````python
if age >= 18:
    print("you can vote!")
````

````python
age = int(input("Enter your age: "))
````

### 2.2. `elif`

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
The <code>elif</code> structure, short for "else if," allows you to define additional conditions. If the initial <code>if</code> condition is <code>False</code>, the program checks each <code>elif</code> condition in order, and if one is <code>True</code>, it executes the corresponding block.<br><br>

```python
if condition1:
    # Code to execute if condition1 is true
elif condition2:
    # Code to execute if condition2 is true
else:
    # Code to execute if neither condition1 nor condition2 is true
```
</div>

````python
score = int(input())
````

````python
if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >=70:
    print("C")
else:
    print("D")
````

## 6. Nested Conditions

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
When using nested conditions, the first condition is evaluated initially. If it is <code>True</code>, the program checks the next condition, and so forth. Finally, the code inside the last condition is executed only if all preceding conditions are met.<br><br>

```python
if condition1:
    if condition2:
        # Code to execute if both condition1 and condition2 are true
    else:
        # Code to execute if condition1 is true but condition2 is false
else:
    # Code to execute if condition1 is false
```
</div>

## 8. `match-case`

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
The <code>match-case</code> structure in Python allows you to check conditions using different patterns and execute corresponding actions. This approach makes the code more readable and avoids long <code>if-elif-else</code> chains.<br><br>

```python
match variable:
    case pattern1:
        # Code to execute if variable matches pattern1
    case pattern2:
        # Code to execute if variable matches pattern2
    case _:
        # Code to execute if no pattern matches (default case)
```

- <code>match</code>: The keyword that looks for a variable to evaluate and match.
- <code>case</code>: Each case specifies a pattern or condition, and if it matches the variable, the code inside it is executed.
- <code>_</code>: The default case, executed if none of the patterns match the variable's value.
</div>


#### 💠Using match-case in Python for days of the week

````python
day = input()
````

````python
match day:
    case "Mon":
        print("Monday")
    case "Wed":
        print("Midday")
    case "Fri":
        print("end of week")
    case "Sat" | "Sun":
        print("weekend")
    case _:
        print("NOT!!!")
````

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
The variable <code>day</code> is selected for evaluation.<br><br>
- Each <code>case</code> matches a specific day.<br>
- <code>case "Saturday" | "Sunday"</code> is a multi-match case and executes when <code>day</code> is either "Saturday" or "Sunday".<br>
- <code>case _</code> acts as the default case and executes if <code>day</code> does not match any of the specified cases.<br><br>

```python
match day:
    case "Monday":
        # Code to execute if day is "Monday"
    case "Saturday" | "Sunday":
        # Code to execute if day is "Saturday" or "Sunday"
    case _:
        # Code to execute if day doesn't match any cases
```
</div>

# Summary of Common String Methods

| Method                | Description                                    | Example                                |
|-----------------------|------------------------------------------------|----------------------------------------|
| `len(text)`           | Returns the length of the string               | `len("Python")` -> `6`                |
| `text.upper()`        | Converts string to uppercase                   | `"python".upper()` -> `PYTHON`        |
| `text.lower()`        | Converts string to lowercase                   | `"PYTHON".lower()` -> `python`        |
| `text.strip()`        | Removes whitespace from the beginning/end      | `" hello ".strip()` -> `hello`        |
| `text.replace(a, b)`  | Replaces substring `a` with `b`                | `"Hello".replace("e", "a")` -> `Hallo` |
| `text.split(delim)`   | Splits string by delimiter into a list         | `"a,b,c".split(",")` -> `['a', 'b', 'c']` |
| `",".join(list)`      | Joins list elements into a single string       | `",".join(['a', 'b'])` -> `a,b`       |
| `text.find(sub)`      | Returns index of first occurrence of substring | `"hello".find("e")` -> `1`            |
| `text.count(sub)`     | Counts occurrences of substring                | `"hello".count("l")` -> `2`           |
| `text.startswith(a)`  | Checks if string starts with substring `a`     | `"hello".startswith("he")` -> `True`  |
| `text.endswith(a)`    | Checks if string ends with substring `a`       | `"hello".endswith("lo")` -> `True`    |
