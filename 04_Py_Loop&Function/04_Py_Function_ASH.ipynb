## 1. Function definition
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">

**Concept of a Function**: A function is a set of instructions designed to perform a specific task or calculate a particular result. Functions allow us to encapsulate a specific operation in code, enabling us to perform that operation by calling the function without the need to repeat the code.<br>

**Role of Functions in Coding**: Functions help us keep our code organized and tidy, reduce code duplication, and make it more understandable. When we use functions, we can easily test and debug each function individually.
</iv>

## 3. Naming Conventions for Functions
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
In Python, proper naming of functions is very important, as the name of the function should reflect its purpose. Here are some principles and standards for naming functions:<br>

- **Descriptive Names**: The name of the function should be descriptive and indicate what the function does. For example, <code>calculate_sum</code> is better than <code>cs</code>, as it clearly shows what the function does.<br>
- **Lowercase Naming Rules and Use of Underscore `_`**: It is recommended to use lowercase letters for naming functions in Python, and if multiple words are needed, separate them with an underscore. This style is known as <i>snake_case</i>.
  
```python
def calculate_total_price():
    # Code to execute for each item
    pass
```
</div>

````python
def calulate_area(W, h):
    """
    This function calculates the area of a rectangle.
    Parameters:
    width (float): The width of the rectangle
    height (float): The height of the rectangle
    Returns:
    float: The area of the rectangle
    """
    return float(W * h)
````

````python
calulate_area(10, 20)
````

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">Point:
    

- **Avoid Using Keywords and Generic Names**: Refrain from using Python keywords (such as <code>def</code>, <code>class</code>, <code>return</code>) or generic names (like <code>sum</code>, <code>print</code>) as function names, as this may cause conflicts or reduce code readability.<br>
- **Avoid Using Abbreviations and Meaningless Names**: Short and meaningless names (like <code>f</code> or <code>x1</code>) can lead to confusion. Try to use names that accurately convey the function's purpose.

## 9. Recursive Functions in Python
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">

- **Concept of Recursion**: Recursion occurs when a function calls itself to break down a complex problem into smaller, more manageable issues. In these types of functions, with repeated calls, a portion of the problem is solved each time until a base case is reached, at which point the computations stop.<br>
- **Functionality of Recursion**: In recursive functions, a base case must be defined to prevent infinite calls. Each time the function is called, it operates with smaller inputs until it reaches this base case. When the base case is satisfied, the function no longer calls itself and begins returning results.
</div>

#### 💠Calculating Factorial
- **Factorial of a Number**: The factorial of a positive number (n!) is defined as the product of all integers from 1 to n. It can be defined as follows:<br>
  
  $$n! = n \times (n - 1) \times (n - 2) \times ... \times 1$$
  
  or recursively as:
  
  $$n! = n \times (n - 1)!$$

  with the base case:
  
  $$0! = 1$$

````python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)
````

````python
factorial(5)
````

#### 💠Fibonacci Series
    
- **Fibonacci Series**: The Fibonacci series is defined as follows:<br>

  $$F(n) = F(n - 1) + F(n - 2)$$

  with the base cases:

  $$F(0) = 0$$  and  $$F(1) = 1$$

````python
def fibo(n):
    if n == 0:
        return 0
    elif n== 1:
        return 1
    else:
        return fibo(n-1) + fibo(n-2)
````

````python
fibo(5)
````

````python
fibo(2)
````

````python
fibo(6)
````
