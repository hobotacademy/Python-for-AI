## 1. `for` Loop

### 1.1. Intro to the for Loop
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
The <code>for</code> loop in Python is used to iterate over a sequence of data, such as lists, strings, and any iterable object. For example, if we have a list of fruits and want to display each fruit, instead of writing the print statement multiple times, we can use a <code>for</code> loop. The general structure of a <code>for</code> loop is as follows:<br><br>

```python
for item in iterable:
    # Code to execute for each item
```</div>


#### 💠Iterating Over a List

````python
fruits = ["apple", "banana", "cherry"]
````

````python
fruits
````

````python
for fruit in fruits:
    print(fruit)
````

#### 💠Using range in a `for` Loop
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
Sometimes, we want the loop to iterate a specific number of times. We can use the <code>range()</code> function for this purpose. For example, if we want to print the numbers from 1 to 10:<br><br>

```python
for number in range(1, 11):# codember)
```</div>


````python
for num in range(1, 11):
    print(num)
````

````python
list(range(1, 11))
````

#### 💠Printing Even Numbers with for
Use a for loop to print the even numbers between 1 and 10. You can use a simple condition to display only the even numbers.

````python
for num in range(1, 101):
    if num%2 == 0:
        print(num, end="\t")
````

## 2. `while` Loop`

### 2.1. Definition of the `while` Loop
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
The <code>while</code> loop continues to execute as long as a certain condition is true. This loop is useful when we do not know exactly how many times we need to repeat a block of code and just want to keep going until a specific condition is met. The general structure of a <code>while</code> loop is:<br><br>

```python
while condition:
    # Code to execute as long as the condition is true
```
</div>


#### 💠Counting to 5 with `while`

````python
counter = 0
````

````python
while counter < 5:
    print(counter)
    counter += 1
````

#### 💠Getting User Input with `while`
Suppose we want to continuously take input from the user until they type "exit." We can use a while loop for this:

````python
user_input = ""
````

````python
while user_input != "exit":
    user_input = input()
````

#### 💠Counting Input Instances
Write a while loop that continues to take input from the user and counts how many times the user has provided input until they type "exit."

````python
counter = 0
````

````python
while True:
    user_input = input()
    if user_input.lower() == "exit":
        break
    counter +=1
    print(counter)
````

## 3. Control Flow in Loops (break and continue)

### 3.1. Using `break`
Sometimes, we may want to exit a loop early. We can achieve this using the break statement. break allows us to terminate the loop before it has completed all its iterations. For example:

````python
for i in range(10):
    if i == 5:
        break
    print(i)
````

````python
list(range(10))
````

### 3.2. Using `continue`
The continue statement is used to skip the current iteration and move to the next one. For example, if we want to print only even numbers:

````python
for i in range(10):
    if i%2 !=0:
        continue
    print(i)
````
