```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("tips.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
missing_data = df.isnull().sum() / len(df) * 100
missing_data = missing_data[missing_data > 0]
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(x=missing_data.index, y=missing_data.values)
plt.title('Missing Data Percentage per Column')
plt.xlabel('Columns')
plt.ylabel('Percentage of Missing Values (%)')
plt.xticks(rotation = 45)
plt.show()
```

```python
num_cols = df.select_dtypes(include=['float64', 'int64']).columns
df[num_cols].hist(figsize=(14, 10), bins=30)
plt.show()
```

```python
# axes.flatten()
```

```python
n_cols = len(num_cols)
cols = 2
rows = (n_cols // cols) + (n_cols % cols > 0)

fig, axes = plt.subplots(rows, cols, figsize=(14, 8))

axes = axes.flatten()

for i, c in enumerate(num_cols):
    sns.boxplot(data=df, x=c, ax=axes[i])
    axes[i].set_title(f'Boxplot of {c}')
    axes[i].tick_params(axis='x', rotation=90)

for j in range(i+1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```
