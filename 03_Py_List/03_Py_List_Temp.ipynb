## 1. Introduction to `Lists`

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">

**Lists in Python**<br>
Lists are one of the structured data types in Python, allowing us to store a collection of elements within a single structure. These elements can be of any data type, such as numbers, strings, or even other lists.<br><br>

**Key Features of Lists**<br>
- **Mutability**: Unlike strings and tuples, lists in Python are mutable, meaning we can modify their elements after creation.<br>
- **Order Preservation**: The order of elements in lists is preserved, allowing access to each element based on its position (index).<br>
- **Data Type Flexibility**: A list can contain different types of data simultaneously.

## 3. Using Slicing to Access a Subset of Elements

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
Slicing allows us to extract a portion of the list using three main parameters: <code>start</code>, <code>stop</code>, and <code>step</code>. The general slicing format is <code>list[start:stop:step]</code>.
</dv>


### 3.1. Basic Slicing

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
This is done by using <code>start</code> and <code>stop</code> (without <code>step</code>). <code>start</code> is the first index included, and <code>stop</code> is the first index excluded.
</div>


### 3.2. Slicing with Negative Indices

### 3.3. Slicing without Specifying start and stop

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
If we leave out <code>start</code> or <code>stop</code>, Python assumes the beginning or end of the list:<br>
- Without <code>start</code>: Slices from the beginning.<br>
- Without <code>stop</code>: Slices to the end.
</div>

### 3.4. Slicing with `step`

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
<code>step</code> specifies the interval between elements selected. For example, a <code>step</code> of 2 will select every second element.

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">

## 4. List Methods

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">

**Detailed Explanation of <code>append()</code>, <code>extend()</code>, <code>insert()</code>, <code>remove()</code>, and <code>pop()</code> Methods in Python Lists**
</div>

### 4.1. Method `append(item)`

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
This method adds a new item to the end of the list. Note that it adds only one item at a time and does not allow adding multiple items simultaneously.
</di>


<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
In this example, the number <code>4</code> has been added to the end of the <code>numbers</code> list.
</di>
>


### 4.2. Method `extend(iterable)`

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
This method is used to add multiple items to the end of the list. It effectively adds another iterable (like another list) to the end of the list.
</di>


<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
In this example, the list <code>[4, 5, 6]</code> has been added to the end of the <code>numbers</code> list.
</di>


### 4.3. Method `insert(index, item)`

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
This method adds an item at a specific position in the list (based on the index). It allows you to specify the exact position of the item.
</div>


<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
In this example, the number <code>9</code> has been added at index <code>1</code> in the list, and the remaining items have been shifted to the right.
</di>


### 4.4. Method `remove(item)`

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
This method removes the first occurrence of an item with a specified value in the list. If the value appears multiple times in the list, only the first instance will be removed. If the value is not found in the list, it will raise an error.
</di>


<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
In this example, the first occurrence of <code>'banana'</code> has been removed from the list.
</di>


### 4.5. Method `pop(index)`

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
This method removes an item from the list and returns its value. If no index is specified, the last item of the list will be removed. If an index is specified, the item at that position will be removed.
</di>


#### 💠Without index

#### 💠With specified index

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
In the first example, the last item is removed, and in the second example, the item at index <code>1</code>, which is <code>2</code>, is removed and assigned to the variable <code>item</code>.

### 4.6. Method `clear()`

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
This method removes all items from the list and converts it into an empty list. This method is suitable when you want to completely clear the contents of a list.
</di>
>


### 4.7. Method `index(item, start, end)`

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
This method returns the first index of a specific element in the list. The search range can be limited using the <code>start</code> and <code>end</code> parameters. If the desired value is not found, it raises an error.
</di>


#### 💠with search range

### 4.8. Method `count(item)`

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
This method returns the number of occurrences of a specific item in the list.
</div>

### 4.9. Method `sort(reverse=False)`
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
This method sorts the list in ascending order. If you pass the parameter <code>reverse=True</code>, the list will be sorted in descending order.
</dv>


#### 💠descending sort

### 4.10. Method `reverse()`
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
This method reverses the order of items in the list without sorting them. The current order of elements will be from the last to the first.
</di>


### 4.11. Method `copy()`
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
This method creates a shallow copy of the list. Changes made to the copied list will not affect the original list.
</di>


<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
In this example, the changes made to <code>copied_list</code> do not affect <code>original_list</code>, as an independent copy has been created.
</di>


## 5. List Functions

### 5.1. Function `len(list)`
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
This function returns the total number of elements in the list and is used to determine the size or length of the list.
</di>


<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
In this example, the number of elements in the <code>numbers</code> list is <code>4</code>.
</di>


### 5.2. Function `sum(list)`
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
This function calculates and returns the sum of all numerical elements in the list. If the elements of the list are not numeric, it will raise an error.
</di>


### 5.3. Function `min(list)`
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
This function returns the smallest value present in the list. All elements of the list must be comparable.
</di>


### 5.4. Function `max(list)`
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
This function returns the largest value present in the list. All elements of the list must be comparable.
</di>


### 5.5. Function `any(list)`
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
This function checks whether at least one of the elements in the list is true (<code>True</code>). If at least one element is true, it returns <code>True</code>; otherwise, it returns <code>False</code>. Values like zero, <code>None</code>, an empty string <code>""</code>, and an empty list <code>[]</code> are considered <code>False</code>.
</di>


### 5.6. Function `all(list)`
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
This function checks whether all the elements in the list are true (<code>True</code>). If all elements are true, it returns <code>True</code>; if even one of them is <code>False</code>, it will return <code>False</code>.
</di>


#### 💠with a False element

### 6.1. `map(function, iterable)`
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
This function applies a given function to each element of an iterable (like a list) and returns a map object. It is very useful for processing groups of elements in a list. You can convert the output to a list using <code>list(map(...))</code>.
</di>


### 6.2. `filter(function, iterable)`
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
This function filters elements that meet a specific condition and returns a filter object. You can also convert the output to a list.
</di>


### 6.3. `reduce(function, iterable)`
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
This function is obtained from the <code>functools</code> module and performs a cumulative operation on the elements of the list, returning a single value.
</di>


### 6.4. `zip(*iterables)`
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
This function combines multiple iterables (like lists) into pairs. It is useful for working with lists and transforming them into related pairs or groups.
</di>


### 6.5. `enumerate(iterable, start=0)`
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
This function returns both the index and the value of each element as a tuple while iterating over a list.
</di>


### 6.6. List Comprehension
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
This is a concise way to create new lists using expressions inside brackets.
</di>


### 6.7. `set(list)`
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
By converting a list to a set, you can remove duplicate elements and then convert it back to a list if needed.
</di>
