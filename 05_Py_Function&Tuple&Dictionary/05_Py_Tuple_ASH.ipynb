# Tuples
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">

**Tuples** are actually a type of ordered data structure that can contain a collection of values. Each element in a tuple is identified by an index, and we can access the elements through these indexes.<br><br>
The uses of tuples in Python include the following:<br>
- **Data Immutability**: Due to their immutability, tuples are used to store data that should not change, such as geographical coordinates (latitude and longitude).<br>
- **Faster Processing**: Tuples are typically faster to process than lists due to their immutability, which eliminates the need for change operations.<br>
- **Used as Dictionary Keys**: Because of their immutability, tuples can be used as keys in dictionaries, while lists cannot.
</div>

## 1. Definition and Structure of Tuples
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">

Tuples are defined simply using parentheses (<code>()</code>), and their elements are separated by commas.<br><br>

Example of a tuple:
```python
my_tuple = (1, 2, 3, 4)
```

````python
t2 = (1, 2, [9, 8, 7], 3)
t2
````

````python
t3 = (1, 2, (5, 6), 3)
t3
````

## 2. Definition of a Single-Element Tuple
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
>
When a tuple is defined with only one element, a comma must be placed after the element for Python to recognize it as a tuple. Otherwise, Python will treat it as a simple value (not a tuple).<br><br>

Example of a single-element tuple:
```python
single_element_tuple = (5,)
```
</div>


````python
non_t = (5)
type(non_t)
````

````python
t4 = (5,)
type(t4)
````

````python
print(type(t4))
````

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">

<code>single_element_tuple</code> is a tuple, and because of the comma after 5, Python recognizes it as a tuple.<br><br>
In contrast, <code>not_a_tuple</code> is considered an integer (int) because there is no comma.<br>

Example:

## 3. Accessing Elements in Tuples
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
    
In tuples, just like lists, we can use indexes to access elements. In Python, indexes start from zero. Additionally, we can use slicing to access a specific section of the tuple.<br>

Example of accessing elements in a tuple:
```python
my_tuple = (10, 20, 30, 40, 50)
print(my_tuple[0])  # Accessing the first element (10)
print(my_tuple[1:4])  # Accessing a slice from index 1 to 3 (20, 30, 40)
```

## 4. Features of Tuples and Differences from Lists
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">

Both tuples and lists are ordered data structures that can store a collection of values, but there are important differences between them that make each more suitable for specific use cases.<br>

**Key Difference**: 
- **Immutability of Tuples**: One of the main differences between tuples and lists is the immutability of tuples. Once a tuple is created, its elements cannot be modified, added, or removed. In contrast, lists are mutable, meaning their elements can be easily changed, removed, or new elements can be added.
</div>

````python
list1 = [1, 2, 3]
tuple1 = (1, 2, 3)
````

````python
list1[0] = 10
list1
````

| Mutability                | Immutable         | Mutable                   |
| Performance               | Higher performance| Lower performance compared to tuple |

## 5. Slicing on Tuples
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">

Slicing is a method in Python that allows us to extract a specific section of a tuple by specifying a range of indexes. The slicing structure is as follows:<br><br>

```python
tuple[start:end:step]
```
Where:
- **start**: The index where the slice starts (inclusive).
- **end**: The index where the slice ends (exclusive).
- **step**: The step or interval between each element in the slice (optional).<br>
</div>

````python
my_tuple
````

````python
my_tuple[:3]
````

````python
my_tuple[::2]
````

### Negative Indexing and Slicing in Tuples
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">

- **Negative Indexing**: In negative indexing, indexing starts from the end of the tuple. For example, -1 accesses the last element of the tuple.<br>
- **Slicing**: The ending index in slicing is exclusive. For example, <code>my_tuple[1:4]</code> will return elements between indexes 1 and 3.<br>
- **Step**: The <code>step</code> allows us to select elements at regular intervals.<br><br>
</div>

````python
my_tuple
````

````python
my_tuple[-1]
````

````python
my_tuple[-3:-1]
````
