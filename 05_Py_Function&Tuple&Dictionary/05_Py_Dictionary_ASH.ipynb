## 1. Definition of Dictionaries
<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">

**Dictionaries** are one of the essential data structures in Python, designed for fast data storage and access. In dictionaries, data is stored in key-value pairs.<br>
In this structure, each key is unique and points to a specific value. In other words, a dictionary allows us to store data in such a way that each value is identifiable by a key. This feature is particularly useful when we need to store data that is quickly searchable.
</div>


```python
students = {
    "Ali":85,
    "sara":90,
    "Reza":65
}
```

```python
type(students)
```

```python
students["Ali"]
```

```python
d = {}
type(d)
```

```python
d2 = dict()
type(d2)
```

```python
d2
```

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">

**Note**: In dictionaries, the order of elements does not need to be preserved because data access is done through keys. However, in lists and tuples, the order of elements matters.
</div>

## 2. Methods

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">

The <code>get()</code> method allows us to retrieve the value associated with a key. If the key does not exist, we can also specify a default value to be returned instead of raising an error.<br>
</div>


```python
students
```

```python
students.get("Sara")
```

```python
students.get("R", "Not")
```

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">

The <code>keys()</code> method returns all the keys in a dictionary as an iterable object, which we can loop over or convert into a list.<br>
</div>

```python
students
```

```python
students.keys()
```

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">

The <code>values()</code> method returns all the values in a dictionary as an iterable object.<br>
</div>>


```python
students.values()
```

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">
The <code>items()</code> method returns all key-value pairs as an iterable object containing tuples. This method is very useful for iterating over dictionaries.
</di>


```python
students.items()
```

```python
for s, g in students.items():
    print(f"{s}: {g}")
```

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">

The <code>pop()</code> method removes a specified key from the dictionary and returns its value. If the key does not exist, we can provide a default value, or without a default, an error will be raised.
</di>


```python
students.pop("Mina")
```

```python
students
```

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">

The <code>popitem()</code> method removes the last key-value pair from the dictionary and returns it as a tuple. In newer versions of Python, this method removes the last item.
</div>


```python
students.popitem()
```

```python
students
```

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">

The <code>clear()</code> method removes all contents of the dictionary, making it empty.
</di>


```python
students.clear()
students
```

```python
del students
# students
```

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">

The <code>len()</code> function returns the number of keys (or key-value pairs) in the dictionary.
</di>


```python
student_grades
```

```python
len(student_grades)
```

<div style="background-color: #dafbe1; border-left: 5px solid #34c759; padding: 10px;">

The <code>in</code> operator can be used to check whether a key exists in a dictionary.
</di>


```python
"Ali" in student_grades
```

```python
104 in student_grades
```
